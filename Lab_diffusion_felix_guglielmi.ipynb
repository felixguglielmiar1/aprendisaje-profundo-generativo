# **Objetivo 2**: Modelo de Difusión para Datos Tabulares



```python
# ======================================================
# Task 2: Diffusion Model for Tabular Data
# ======================================================

# ------------------------------------------------------
# PARTE 1: (Forward Diffusion Schedule)
# ------------------------------------------------------

class DiffusionSchedule:
    def __init__(self, timesteps=100, beta_start=1e-4, beta_end=0.02, device="cpu"):
        self.timesteps = timesteps # Cantidad total de pasos de RUIDO (T)
        self.device = device

        # 'betas' RUIDO agregado en cada paso t.
        # Empieza (0.0001) y sube linealmente hasta 0.02.
        self.betas = torch.linspace(beta_start, beta_end, timesteps, device=device)

        # 'alphas' es cuánto del dibujo original se conserva en ese paso individual (1 - beta).
        self.alphas = 1.0 - self.betas

        # 'alpha_hat' (o alpha_bar) es el producto acumulado:
        #  dice cuánto del dibujo original queda intacto desde el paso 0 hasta el paso 't' directamente.
        # si queremos ver cómo queda la foto en el paso 50,
        # no hace falta agregar ruido 50 veces en cada bucle, Saltamos directo al paso 50 con una fórmula.
        self.alpha_hat = torch.cumprod(self.alphas, dim=0)

    def add_noise(self, x_0, t):
        """
        Esta función toma el dato limpio (x_0) y el número de paso (t),
        y nos devuelve el dato con ruido (x_t) junto con el (noise) agregado.
        """
        # 1. Creamos ruido al azar (ruido Gaussiano con media 0 y varianza 1) del mismo tamaño que x_0
        noise = torch.randn_like(x_0)

        # 2. Buscamos  cuánto del dibujo original queda en el paso 't' (sqrt_alpha_hat)
        # y cuánta ruido toca poner en el paso 't' (sqrt_one_minus_alpha_hat)
        sqrt_alpha_hat = torch.sqrt(self.alpha_hat[t]).unsqueeze(1)
        sqrt_one_minus_alpha_hat = torch.sqrt(1.0 - self.alpha_hat[t]).unsqueeze(1)

        # 3. Mezclamos el dato limpio con el noise según la receta del paso 't':
        # Fórmula oficial: x_t = sqrt(alpha_hat_t) * x_0 + sqrt(1 - alpha_hat_t) * noise
        x_t = sqrt_alpha_hat * x_0 + sqrt_one_minus_alpha_hat * noise

        return x_t, noise


# ------------------------------------------------------
# PARTE 2: (Red Neuronal MLP)
# ------------------------------------------------------
# (la red neuronal) recibe el dato sucio (x_t) y predice:
# "¿cuanto noise se le agrego al dato?".
# para la prediccion se necesita saber en qué paso de agregacion de noise estamos (no es lo mismo
# limpiar algo del paso 2 que algo del paso 99). Para eso sirve el Time Embedding (el reloj).

class SinusoidalTimeEmbedding(nn.Module):
    """
    Convierte un número entero (ej: paso t=45) en un vector de números (ondas seno y coseno)
    para que la red neuronal entienda fácilmente en qué momento del tiempo está.
    """
    def __init__(self, dim):
        super().__init__()
        self.dim = dim

    def forward(self, t):
        half_dim = self.dim // 2
        emb = math.log(10000) / (half_dim - 1)
        emb = torch.exp(torch.arange(half_dim, device=t.device) * -emb)
        emb = t[:, None].float() * emb[None, :]
        emb = torch.cat((emb.sin(), emb.cos()), dim=-1)
        return emb


class DenoisingMLP(nn.Module):
    """
       Una red neuronal para datos tabulares (MLP).
    - Entrada: Los datos con noise (x_t) + el paso en el tiempo (t).
    - Salida: noise (ruido) que predicho que contienen los datos.
    """
    def __init__(self, input_dim, hidden_dim=128, time_dim=32):
        super().__init__()
        # transforma 't' en un vector de tamaño time_dim
        self.time_mlp = nn.Sequential(
            SinusoidalTimeEmbedding(time_dim),
            nn.Linear(time_dim, time_dim),
            nn.SiLU() # Función de activación suave (como ReLU pero mejor para difusión)
        )

        # recibe (datos con noise + vector de tiempo) pegados juntos
        self.net = nn.Sequential(
            nn.Linear(input_dim + time_dim, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, input_dim) # Devuelve un vector del mismo tamaño que el dato original
        )

    def forward(self, x_t, t):
        # 1. Miramos el reloj
        t_emb = self.time_mlp(t)
        # 2. Pegamos el dato noise con la información del reloj
        x_input = torch.cat([x_t, t_emb], dim=1)
        # 3. Predecimos el noise (ruido)
        predicted_noise = self.net(x_input)
        return predicted_noise


# ------------------------------------------------------
# PARTE 3: Entrenamiento (SOLO con transacciones normales)
# ------------------------------------------------------
# ¿Cuál es el truco para detectar fraudes?
# Le enseñamos  a limpiar SOLAMENTE transacciones normales (legítimas).
# la idea es que el error de prediccion sea muy alto en las transacciones que no son normales

input_dim = X_train_scaled.shape[1]
TIMESTEPS = 100
BATCH_SIZE = 512
EPOCHS = 20
LR = 1e-3

diff_schedule = DiffusionSchedule(timesteps=TIMESTEPS, device=device)
model = DenoisingMLP(input_dim=input_dim, hidden_dim=128, time_dim=32).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=LR)

# (DataLoader)
train_dataset = TensorDataset(torch.tensor(X_train_scaled, dtype=torch.float32))
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)

loss_history = []

print("Entrenando el modelo de difusión (solo con datos normales)...")
model.train()
for epoch in range(EPOCHS):
    epoch_loss = 0.0
    for (x_0_batch,) in train_loader:
        x_0_batch = x_0_batch.to(device)

        # 1. Elegimos un paso de tiempo 't' al azar para cada transacción del lote (entre 0 y 99)
        t = torch.randint(0, TIMESTEPS, (x_0_batch.size(0),), device=device).long()

        # 2. agregamos ruido hasta ese paso 't'
        x_t, true_noise = diff_schedule.add_noise(x_0_batch, t)

        # 3. se predice el ruido
        predicted_noise = model(x_t, t)

        # 4. Calculamos el error
        loss = F.mse_loss(predicted_noise, true_noise)

        # 5. (Backpropagation)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item() * x_0_batch.size(0)

    epoch_loss /= len(train_dataset)
    loss_history.append(epoch_loss)
    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f"Epoch [{epoch+1}/{EPOCHS}] - Loss (MSE del Ruido): {epoch_loss:.4f}")

# Graficamos la curva de aprendizaje
plt.figure(figsize=(7, 4))
plt.plot(range(1, EPOCHS + 1), loss_history, marker='o', color='#2ecc71')
plt.title('Curva de Entrenamiento del Modelo de Difusión')
plt.xlabel('Época')
plt.ylabel('MSE Loss (Error prediciendo el ruido)')
plt.grid(True, alpha=0.3)
plt.show()
```

# **Objetivo 3**: Puntaje de Anomalía basado en Difusión (Anomaly Scoring)

```python
# ======================================================
# Task 3: Diffusion-Based Anomaly Scoring
# ======================================================
# Tomamos una transacción sospechosa (x_0), le agregamos noise en un paso 't', y le pedimos a la red que prediga el ruido.
# - Si la transacción es NORMAL: -> Error BAJO.
# - Si la transacción es FRAUDE: --> Error ALTO.
# ¡Ese error es nuestro Puntaje de Anomalía (Anomaly Score)!

@torch.no_grad() # Apagamos el aprendizaje
def compute_diffusion_anomaly_scores(model, schedule, X_data, n_samples=20, score_type="residual", batch_size=1024):
    """
    Calcula qué tan difícil es para el modelo limpiar cada transacción.
    - score_type='residual': Mide el error al predecir el noise.
    - score_type='reconstruction': Mide el error al reconstruir el dato limpio original (x_0_hat).
    - n_samples: Cuántos pasos 't' distintos probamos por cada dato para promediar (Stochastic Averaging).
    """
    model.eval()
    dataset = TensorDataset(torch.tensor(X_data, dtype=torch.float32))
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=False)

    all_scores = []

    for (x_0_batch,) in loader:
        x_0_batch = x_0_batch.to(device)
        batch_scores = torch.zeros(x_0_batch.size(0), device=device)

        # Promedio Estocástico:
        # Probamos 'n_samples' veces (ej. 20 veces en distintos pasos t) y promediamos el error.
        for _ in range(n_samples):
            # Elegimos un paso 't' al azar (evitamos los extremos muy cercanos a 0 o a T para mayor estabilidad)
            t = torch.randint(10, schedule.timesteps - 10, (x_0_batch.size(0),), device=device).long()

            # Ensuciamos el dato
            x_t, true_noise = schedule.add_noise(x_0_batch, t)

            # predice el ruido
            pred_noise = model(x_t, t)

            if score_type == "residual":
                # Opción A (Residual): Diferencia directa entre el ruido real y el ruido predicho
                error = torch.mean((true_noise - pred_noise) ** 2, dim=1)
            else:
                # Opción B (Reconstrucción): Despejamos la ecuación para ver cómo quedaría el dato limpio (x_0_hat)
                sqrt_alpha_hat = torch.sqrt(schedule.alpha_hat[t]).unsqueeze(1)
                sqrt_one_minus_alpha_hat = torch.sqrt(1.0 - schedule.alpha_hat[t]).unsqueeze(1)
                x_0_hat = (x_t - sqrt_one_minus_alpha_hat * pred_noise) / sqrt_alpha_hat
                error = torch.mean((x_0_batch - x_0_hat) ** 2, dim=1)

            batch_scores += error

        # Dividimos por la cantidad de intentos para tener el promedio limpio
        batch_scores /= n_samples
        all_scores.append(batch_scores.cpu().numpy())

    return np.concatenate(all_scores)

# Calculamos los puntajes en Validación y Test
print("Calculando puntajes de anomalía en Validación y Test...")
val_scores_diff = compute_diffusion_anomaly_scores(model, diff_schedule, X_val_scaled, n_samples=25, score_type="residual")
test_scores_diff = compute_diffusion_anomaly_scores(model, diff_schedule, X_test_scaled, n_samples=25, score_type="residual")
print("¡Listo! Ejemplo de primeros 5 puntajes:", val_scores_diff[:5])
```

# **Objetivo 4**: Evaluación sin Umbral (AUROC y AUPRC)

# **Objetivo 5**: Selección del Umbral (Threshold Selection) en Validación
Hasta ahora el modelo nos da un número continuo (ej: 1.45, 12.8). Necesitamos elegir un número de corte (umbral) usando solamente el conjunto de Validación para decidir a partir de qué valor bloqueamos la tarjeta

```python
# ======================================================
# Task 5: Threshold Selection (Validation Set ONLY)
# ======================================================

precisions, recalls, thresholds = precision_recall_curve(y_val, val_scores_diff)
# Quitamos el último elemento de precision y recall (que siempre es 1 y 0 artificialmente en sklearn)
precisions, recalls = precisions[:-1], recalls[:-1]

# Estrategia 1: Maximizar F1-Score (Balance ideal entre Precision y Recall)
f1_scores = (2 * precisions * recalls) / (precisions + recalls + 1e-8)
best_f1_idx = np.argmax(f1_scores)
thresh_max_f1 = thresholds[best_f1_idx]

# Estrategia 2: Target Recall >= 80% (Queremos atrapar al menos el 80% de los fraudes aunque haya más falsas alarmas)
target_recall = 0.80
valid_recall_indices = np.where(recalls >= target_recall)[0]
thresh_target_rec = thresholds[valid_recall_indices[-1]] # El umbral más alto que cumple >= 80% de recall

print(f"Umbral Estrategia 1 (Max F1 = {f1_scores[best_f1_idx]:.3f}): {thresh_max_f1:.4f}")
print(f"Umbral Estrategia 2 (Recall >= 80%): {thresh_target_rec:.4f}")

# Visualizar Matrices de Confusión en Validación para ambos umbrales
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

for ax, thresh, title in zip(axes, [thresh_max_f1, thresh_target_rec], ["Estrategia 1: Max F1", "Estrategia 2: Target Recall (80%)"]):
    preds = (val_scores_diff >= thresh).astype(int)
    cm = confusion_matrix(y_val, preds)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax, cbar=False,
                xticklabels=['Pred: Normal', 'Pred: Fraude'],
                yticklabels=['Real: Normal', 'Real: Fraude'])
    p = cm[1,1] / (cm[0,1] + cm[1,1] + 1e-8)
    r = cm[1,1] / (cm[1,0] + cm[1,1] + 1e-8)
    ax.set_title(f"{title}\n(Precision: {p:.2f} | Recall: {r:.2f})")

plt.tight_layout()
plt.show()
```

**Objetivo 6**: Evaluación Final en Test
Fijamos el umbral ganador de validación (por ejemplo, thresh_max_f1) y lo aplicamos una única vez sobre los datos de Test.

```python
# ======================================================
# Task 6: Final Test Evaluation
# ======================================================

# Usamos el umbral elegido en Validación (Max F1) sin tocarlo
chosen_threshold = thresh_max_f1
test_preds = (test_scores_diff >= chosen_threshold).astype(int)

cm_test = confusion_matrix(y_test, test_preds)
tn, fp, fn, tp = cm_test.ravel()

precision_test = tp / (tp + fp + 1e-8)
recall_test = tp / (tp + fn + 1e-8)
fpr_test = fp / (fp + tn + 1e-8)
f1_test = f1_score(y_test, test_preds)

print("=== Resultados Finales en Test (Umbral Max-F1 fijado en Validación) ===")
print(f"Precision : {precision_test:.4f}")
print(f"Recall    : {recall_test:.4f}")
print(f"FPR       : {fpr_test:.4f} (Tasa de Falsos Positivos)")
print(f"F1-Score  : {f1_test:.4f}")

plt.figure(figsize=(5, 4))
sns.heatmap(cm_test, annot=True, fmt='d', cmap='Greens', cbar=False,
            xticklabels=['Pred: Normal', 'Pred: Fraude'],
            yticklabels=['Real: Normal', 'Real: Fraude'])
plt.title(f"Matriz de Confusión en Test (F1 = {f1_test:.3f})")
plt.show()
```

# **Objetivo 7**: Comparación con Baselines (Isolation Forest y Autoencoder)
Entrenamos dos modelos clásicos de detección de anomalías usando exactamente el mismo conjunto de entrenamiento (solo normales) y los comparamos en Test.

```python
# ======================================================
# Task 7: Baseline Comparisons
# - Isolation Forest
# - Gaussian Mixture Autoencoder (GM-AE)
# - Normalizing Flow (RealNVP) [con comentarios ELI5 de 5 años]
# - LightGBM [con preparación de dataset y peso de clase minoritaria]
# ======================================================

ensure_pkg("lightgbm")
import lightgbm as lgb

# ------------------------------------------------------
# 7.1 Baseline 1: Isolation Forest (solo normales)
# ------------------------------------------------------
print("Entrenando Isolation Forest...")
iso_forest = IsolationForest(
    n_estimators=200,
    contamination="auto",
    random_state=42,
    n_jobs=-1,
)
iso_forest.fit(X_train_scaled)

# decision_function devuelve valores más altos para normales y más bajos para anómalos.
# Invertimos el signo para que un puntaje más alto = mayor anomalía.
val_scores_if = -iso_forest.decision_function(X_val_scaled)
test_scores_if = -iso_forest.decision_function(X_test_scaled)


# ------------------------------------------------------
# 7.2 Baseline 2: Gaussian Mixture Autoencoder (GM-AE)
# Reemplaza al Autoencoder convencional:
# Aprende a reconstruir transacciones normales Y a ubicar sus códigos
# latentes dentro de una mixtura de K gaussianas en el espacio latente.
# ------------------------------------------------------
class GaussianMixtureAutoencoder(nn.Module):
    def __init__(self, input_dim: int, latent_dim: int = 8, n_components: int = 4, hidden_dim: int = 64):
        super().__init__()
        self.latent_dim = latent_dim
        self.n_components = n_components

        self.encoder = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.SiLU(),
            nn.Linear(hidden_dim // 2, latent_dim),
        )

        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim // 2),
            nn.SiLU(),
            nn.Linear(hidden_dim // 2, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, input_dim),
        )

        # Parámetros de la Mixtura de Gaussianas en el espacio latente:
        # logits de mezcla -> pi_k, medias mu_k (K, D), log_var_k (K, D)
        self.pi_logits = nn.Parameter(torch.zeros(n_components))
        self.means = nn.Parameter(torch.randn(n_components, latent_dim) * 0.5)
        self.log_vars = nn.Parameter(torch.zeros(n_components, latent_dim))

    def encode_decode(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
        z = self.encoder(x)
        x_hat = self.decoder(z)
        return z, x_hat

    def gmm_neg_log_likelihood(self, z: torch.Tensor) -> torch.Tensor:
        """
        Calcula -log p(z) bajo la mixtura de K gaussianas diagonales.
        z: (B, D) -> devuelve (B,)
        """
        log_pi = F.log_softmax(self.pi_logits, dim=0)  # (K,)
        var = torch.exp(self.log_vars).clamp(min=1e-4, max=10.0)  # (K, D)

        # z_exp: (B, 1, D), means: (1, K, D), var: (1, K, D)
        z_exp = z.unsqueeze(1)
        mu_exp = self.means.unsqueeze(0)
        var_exp = var.unsqueeze(0)

        # Log-densidad gaussiana diagonal para cada componente k
        log_prob_k = -0.5 * (
            self.latent_dim * math.log(2 * math.pi)
            + torch.sum(torch.log(var_exp) + ((z_exp - mu_exp) ** 2) / var_exp, dim=2)
        )  # (B, K)

        log_mix = torch.logsumexp(log_pi.unsqueeze(0) + log_prob_k, dim=1)  # (B,)
        return -log_mix

    def anomaly_score(self, x: torch.Tensor, lambda_gmm: float = 0.05) -> torch.Tensor:
        z, x_hat = self.encode_decode(x)
        recon_err = torch.mean((x - x_hat) ** 2, dim=1)
        nll_z = self.gmm_neg_log_likelihood(z)
        return recon_err + lambda_gmm * nll_z


print("Entrenando Gaussian Mixture Autoencoder (solo con datos normales)...")
gm_ae = GaussianMixtureAutoencoder(
    input_dim=input_dim,
    latent_dim=8,
    n_components=4,
    hidden_dim=64,
).to(device)

optimizer_gmae = torch.optim.Adam(gm_ae.parameters(), lr=1e-3)
LAMBDA_GMM = 0.05
EPOCHS_GMAE = 20

gm_ae.train()
for epoch in range(EPOCHS_GMAE):
    for (x_0_batch,) in train_loader:
        x_0_batch = x_0_batch.to(device)
        z, x_hat = gm_ae.encode_decode(x_0_batch)
        recon_loss = F.mse_loss(x_hat, x_0_batch)
        gmm_loss = gm_ae.gmm_neg_log_likelihood(z).mean()
        loss = recon_loss + LAMBDA_GMM * gmm_loss

        optimizer_gmae.zero_grad()
        loss.backward()
        optimizer_gmae.step()


@torch.no_grad()
def compute_gmae_scores(model_gmae, X_data, lambda_gmm=0.05, batch_size=1024):
    model_gmae.eval()
    loader = DataLoader(TensorDataset(torch.tensor(X_data, dtype=torch.float32)), batch_size=batch_size, shuffle=False)
    scores = []
    for (x_batch,) in loader:
        x_batch = x_batch.to(device)
        scores.append(model_gmae.anomaly_score(x_batch, lambda_gmm=lambda_gmm).cpu().numpy())
    return np.concatenate(scores)


val_scores_gmae = compute_gmae_scores(gm_ae, X_val_scaled, lambda_gmm=LAMBDA_GMM)
test_scores_gmae = compute_gmae_scores(gm_ae, X_test_scaled, lambda_gmm=LAMBDA_GMM)


# ------------------------------------------------------
# 7.3 Baseline 3: Normalizing Flow (RealNVP)
# Comentarios de entrenamiento explicados para un niño de 5 años
# ------------------------------------------------------
class CouplingLayer(nn.Module):
    def __init__(self, dim: int, hidden_dim: int = 64, mask_even: bool = True):
        super().__init__()
        self.dim = dim
        mask = torch.zeros(dim, dtype=torch.float32)
        if mask_even:
            mask[::2] = 1.0
        else:
            mask[1::2] = 1.0
        self.register_buffer("mask", mask)

        self.net = nn.Sequential(
            nn.Linear(dim, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.SiLU(),
        )
        self.scale_head = nn.Linear(hidden_dim, dim)
        self.shift_head = nn.Linear(hidden_dim, dim)

    def forward(self, x: torch.Tensor) -> Tuple[torch.Tensor, torch.Tensor]:
        # x -> z (transformación directa: de datos reales a cajita ordenada normal)
        x_masked = x * self.mask
        h = self.net(x_masked)
        s = torch.tanh(self.scale_head(h)) * (1.0 - self.mask)
        t = self.shift_head(h) * (1.0 - self.mask)

        z = x_masked + (1.0 - self.mask) * (x * torch.exp(s) + t)
        log_det_jac = torch.sum(s, dim=1)
        return z, log_det_jac


class TabularNormalizingFlow(nn.Module):
    def __init__(self, dim: int, n_blocks: int = 4, hidden_dim: int = 64):
        super().__init__()
        self.dim = dim
        self.layers = nn.ModuleList([
            CouplingLayer(dim=dim, hidden_dim=hidden_dim, mask_even=(i % 2 == 0))
            for i in range(n_blocks)
        ])

    def log_prob(self, x: torch.Tensor) -> torch.Tensor:
        z = x
        total_log_det = torch.zeros(x.size(0), device=x.device)
        for layer in self.layers:
            z, log_det = layer(z)
            total_log_det = total_log_det + log_det

        # Distribución base: Gaussiana estándar N(0, I)
        log_pz = -0.5 * torch.sum(z ** 2 + math.log(2 * math.pi), dim=1)
        return log_pz + total_log_det


print("Entrenando Normalizing Flow (solo con datos normales)...")
nf_model = TabularNormalizingFlow(dim=input_dim, n_blocks=4, hidden_dim=64).to(device)
optimizer_nf = torch.optim.Adam(nf_model.parameters(), lr=1e-3)
EPOCHS_NF = 20

nf_model.train()
for epoch in range(EPOCHS_NF):
    for (x_0_batch,) in train_loader:
        x_0_batch = x_0_batch.to(device)

        # 1. Imagina que tienes un montón de plastilina revuelta (las compras normales)
        #    y quieres acomodarla dentro de una cajita redondita y súper ordenada (una campana normal).
        # 2. El modelo agarra un puñadito de plastilina (el lote) y la estira o la mueve un poquito
        #    con sus maquinitas mágicas para que encaje bien en el centro de la cajita.
        log_px = nf_model.log_prob(x_0_batch)

        # 3. Si la plastilina queda muy lejos del centro o se aplasta/estira raro, el modelo recibe
        #    un "puntito de regaño" (la pérdida negativa del logaritmo). ¡Queremos que el regaño sea chiquito!
        loss = -log_px.mean()

        # 4. Borramos las pistas viejas de cómo lo hicimos antes...
        optimizer_nf.zero_grad()
        # 5. ...revisamos qué maquinita hay que ajustar un poquito...
        loss.backward()
        # 6. ...y movemos las perillas del modelo para que la próxima vez ordene aún mejor.
        optimizer_nf.step()


@torch.no_grad()
def compute_nf_scores(model_nf, X_data, batch_size=1024):
    model_nf.eval()
    loader = DataLoader(TensorDataset(torch.tensor(X_data, dtype=torch.float32)), batch_size=batch_size, shuffle=False)
    scores = []
    for (x_batch,) in loader:
        x_batch = x_batch.to(device)
        # Menor log_prob = más raro/anómalo -> puntaje de anomalía = -log p(x)
        nll = -model_nf.log_prob(x_batch)
        scores.append(nll.cpu().numpy())
    return np.concatenate(scores)


val_scores_nf = compute_nf_scores(nf_model, X_val_scaled)
test_scores_nf = compute_nf_scores(nf_model, X_test_scaled)


# ------------------------------------------------------
# 7.5 Evaluación en Test de todos los Baselines (mismas métricas)
# ------------------------------------------------------
print("\n=== Resultados en TEST (Baselines) ===")
_, _ = evaluate_scores(y_test, test_scores_if, "Test - Isolation Forest")
_, _ = evaluate_scores(y_test, test_scores_gmae, "Test - Gaussian Mixture Autoencoder")
_, _ = evaluate_scores(y_test, test_scores_nf, "Test - Normalizing Flow (RealNVP)")
```

```python
import lightgbm as lgb
from sklearn.metrics import average_precision_score, roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler


def prepare_and_train_lgbm(
    df_raw: pd.DataFrame,
    target_col: str = "Class",
    test_size: float = 0.20,
    val_size_from_temp: float = 0.20,
    scale_pos_weight_override: float | None = 1.0,
    random_state: int = 42,
):
    X_raw = df_raw.drop(columns=[target_col]).to_numpy(dtype=np.float32)
    y_raw = df_raw[target_col].to_numpy(dtype=int)

    X_trainval_lgb, X_test_lgb, y_trainval_lgb, y_test_lgb = train_test_split(
        X_raw,
        y_raw,
        test_size=test_size,
        stratify=y_raw,
        random_state=random_state,
    )

    X_train_lgb, X_val_lgb, y_train_lgb, y_val_lgb = train_test_split(
        X_trainval_lgb,
        y_trainval_lgb,
        test_size=val_size_from_temp,
        stratify=y_trainval_lgb,
        random_state=random_state,
    )

    scaler_lgb = StandardScaler()
    X_train_lgb_sc = scaler_lgb.fit_transform(X_train_lgb)
    X_val_lgb_sc = scaler_lgb.transform(X_val_lgb)
    X_test_lgb_sc = scaler_lgb.transform(X_test_lgb)

    n_pos_lgb = int((y_train_lgb == 1).sum())
    n_neg_lgb = int((y_train_lgb == 0).sum())
    natural_ratio = (n_neg_lgb / n_pos_lgb) if n_pos_lgb > 0 else 1.0
    spw = natural_ratio if scale_pos_weight_override is None else scale_pos_weight_override

    model = lgb.LGBMClassifier(
        objective="binary",
        n_estimators=500,
        learning_rate=0.05,
        num_leaves=31,
        min_child_samples=50,
        subsample_for_bin=50000,
        scale_pos_weight=spw,
        random_state=random_state,
        n_jobs=-1,
        verbose=-1,
    )

    model.fit(
        X_train_lgb_sc,
        y_train_lgb,
        eval_X=X_val_lgb_sc,
        eval_y=y_val_lgb,
        callbacks=[lgb.early_stopping(stopping_rounds=40, first_metric_only=True, verbose=False)],
    )

    val_scores = model.predict_proba(X_val_lgb_sc)[:, 1]
    test_scores = model.predict_proba(X_test_lgb_sc)[:, 1]

    print(
        f"classes_: {model.classes_} | spw={spw:.2f} (natural={natural_ratio:.2f}) | "
        f"best_iter: {model.best_iteration_}"
    )

    return {
        "model": model,
        "scaler": scaler_lgb,
        "scale_pos_weight": spw,
        "X_val": X_val_lgb_sc,
        "y_val": y_val_lgb,
        "val_scores": val_scores,
        "X_test": X_test_lgb_sc,
        "y_test": y_test_lgb,
        "test_scores": test_scores,
    }


# Probamos con scale_pos_weight=1.0 primero (luego puedes probar None para usar el ratio natural)
lgbm_res = prepare_and_train_lgbm(
    df,
    target_col="Class",
    scale_pos_weight_override=1.0,
    random_state=42,
)
_, _ = evaluate_scores(lgbm_res["y_val"], lgbm_res["val_scores"], "LGBM Val (propio)")
_, _ = evaluate_scores(lgbm_res["y_test"], lgbm_res["test_scores"], "LGBM Test (propio)")
```

#  **¿Qué método funciona mejor aquí?**
El modelo supervisado es el que tiene un mejor rendimiento en el area bajo la curva del presicion y el recall.
